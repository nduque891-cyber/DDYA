El problema consiste en determinar cuántos nuevos suscriptores habrá en un mes determinado siguiendo una relación recursiva tipo Fibonacci donde cada mes depende de los dos meses anteriores por ejemplo el mes 0 tiene 0 suscriptores el mes 1 tiene 1 y a partir de ahí cada valor se calcula como la suma de los dos anteriores este enfoque puede ser costoso si se repite muchas veces por lo que se propone usar memoización para guardar resultados ya calculados y así mejorar la eficiencia del proceso al evitar operaciones redundantes

### 5) Análisis de complejidad **[15 pts]**

#### Justificación

La función utiliza un diccionario llamado memo para guardar los resultados ya calculados. Así, cada valor se calcula una sola vez y no se repiten subproblemas.

#### Función de recurrencia

- Caso base: T(0) = T(1) = Θ(1)
- Caso general:T(n) = T(n-1) + Θ(1)

Por lo tanto, la complejidad temporal es:

T(n) = Θ(n)

#### Complejidad temporal

- Mejor caso: Θ(1), cuando n es 0 o 1.
- Peor caso: Θ(n), cuando se calcula un valor mayor que 1.

#### Complejidad espacial

El diccionario memo almacena hasta n resultados y la recursión puede alcanzar una profundidad de n Por esto, la complejidad espacial es Θ(n)

En conclusión, gracias a la memoización, el algoritmo utiliza Θ(n) tiempo y Θ(n) espacio. Sin memoización, tendría una complejidad exponencial Θ(2^n)


## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def nuevos_suscriptores(n, memo=None):
    if memo is None:
        memo = {}

    if n == 0:
        return 0
    if n == 1:
        return 1

    if n not in memo:
        memo[n] = (nuevos_suscriptores(n - 1, memo)+ nuevos_suscriptores(n - 2, memo))

    return memo[n]


n = int(input("Ingrese el mes n: "))
print(f"Nuevos suscriptores en el mes {n}: {nuevos_suscriptores(n)}")

```
